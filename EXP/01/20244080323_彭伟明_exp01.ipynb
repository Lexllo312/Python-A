Python 的四个运算符含义不同：

- `**` 幂运算。

下面计算并把值和类型一起打印出来。

```python
results = {
    "8 / 3": 8 / 3,
    "8 // 3": 8 // 3,
    "8 % 3": 8 % 3,
    "2 ** 7": 2 ** 7,
}
meaning = {
    "8 / 3": "真除法：得到精确的浮点商",
    "8 // 3": "向下取整除法：商取整，抛弃小数部分",
    "8 % 3": "取模：返回除法的余数",
    "2 ** 7": "幂运算：2 的 7 次方",
}
for expr, value in results.items():
    print(f"{expr:<8} = {value!r:<20} 类型={type(value).__name__:<8} {meaning[expr]}")
```

## 实验2：函数、条件、循环与递归

### 2-1 score_level(score) 分档函数

分档规则：`90~100` 优秀，`80~<90` 良好，`60~<80` 及格，`0~<60` 不及格；超出 `0~100` 抛 `ValueError`。

用「早返回」写法，从上到下依次判断，避免多层嵌套。

```python
def score_level(score):
    """把百分制成绩映射为等级字符串。超出 0~100 范围时抛 ValueError。"""
    if score < 0 or score > 100:
        raise ValueError(f"成绩必须在 0~100 之间，收到 {score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for value in [100, 90, 89.9, 80, 79.9, 60, 59.9, 0]:
    print(f"score={value:<6} → {score_level(value)}")
```

```python
print("边界外的输入应抛 ValueError：")
for bad in [-1, 101]:
    try:
        score_level(bad)
    except ValueError as exc:
        print(f"  score={bad:<5} → ValueError: {exc}")
```

### 2-2 for 与 while 求 1~100 的和

`for` 直接遍历 `range(1, 101)`；`while` 手动维护计数器。两种实现结果应当一致，并与内置 `sum()` 对照。

```python
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for   求和 :", total_for)
print("while 求和 :", total_while)
print("sum() 对照 :", sum(range(1, 101)))
print("三者一致   :", total_for == total_while == sum(range(1, 101)))
```

### 2-3 阶乘：递归版与迭代版

约定 `0! = 1`，输入必须是非负整数。递归版从 `n` 乘到 `1`，迭代版用循环累乘。

对比 `n = 0, 1, 5, 10` 两个版本的结果。

```python
def _check_non_negative_int(n):
    if isinstance(n, bool) or not isinstance(n, int):
        raise ValueError(f"输入必须是非负整数，收到 {type(n).__name__}: {n!r}")
    if n < 0:
        raise ValueError(f"输入必须是非负整数，收到 {n}")


def factorial_recursive(n):
    _check_non_negative_int(n)
    if n <= 1:            # 0! = 1! = 1，递归出口
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    _check_non_negative_int(n)
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


print(f"{'n':>3} | {'递归版':>20} | {'迭代版':>20} | 一致")
print("-" * 62)
for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    print(f"{n:>3} | {r:>20} | {it:>20} | {r == it}")
```

### 2-4 summary(values) 返回统计元组

接收一个**非空**数值列表，以元组返回 `(最小值, 最大值, 平均值)`，调用方用元组解包接收。

空列表直接算平均值会 `ZeroDivisionError`，所以显式抛 `ValueError`。

```python
def summary(values):
    """返回 (最小值, 最大值, 平均值)。values 不能为空。"""
    values = list(values)
    if not values:
        raise ValueError("数值列表不能为空")
    return min(values), max(values), sum(values) / len(values)


sample = [85, 58, 92, 76, 60, 89]

lowest, highest, mean = summary(sample)      # 元组解包接收
print(f"输入     = {sample}")
print(f"最小值 = {lowest}")
print(f"最大值 = {highest}")
print(f"平均值 = {mean}")

print()
print("返回值确实是元组 :", type(summary(sample)).__name__)
print("sum([1, 2, 3, 4]) →", summary([1, 2, 3, 4]))

try:
    summary([])
except ValueError as exc:
    print("空列表 →", f"ValueError: {exc}")
```

### 2-5 sum_all(*args) 可变位置参数

`*args` 把所有位置参数收集成一个元组，未传参时就是空元组，`sum(())` 得 0。

分别测试：无参数、单个参数、多个参数。

```python
def sum_all(*args):
    """返回所有位置参数之和；不传参数时返回 0。"""
    return sum(args)


print("无参数      :", sum_all())
print("单个参数    :", sum_all(10))
print("多个参数    :", sum_all(1, 2, 3, 4, 5))
print("含浮点数    :", sum_all(1.5, 2.5, 3))
print("参数元组    :", (1, 2, 3), "→ 展开传入 →", sum_all(*(1, 2, 3)))
```

### 3-5 别名赋值与浅拷贝的区别

`alias = demo` 只是给同一个列表对象再起一个名字，`alias is demo` 为真；`copied = demo.copy()` 创建了一个新列表。

修改 `alias[0]` 会同时改变 `demo`，但不会影响 `copied`。

> 本题按要求使用**单独的演示列表** `demo`，避免改变后续实验要用的原始成绩 `scores`。

```python
demo = [10, 20, 30]
alias = demo           # 只是别名，指向同一个对象
copied = demo.copy()   # 浅拷贝，产生新对象

print("修改前：")
print(f"  demo   = {demo}")
print(f"  alias  = {alias}")
print(f"  copied = {copied}")

alias[0] = 999         # 通过别名修改第一个元素

print()
print("修改 alias[0] = 999 之后：")
print(f"  demo   = {demo}      ← 跟着变了")
print(f"  alias  = {alias}")
print(f"  copied = {copied}    ← 不受影响")

print()
print("身份比较：")
print(f"  alias  is demo   → {alias is demo}   （同一对象）")
print(f"  copied is demo   → {copied is demo}  （不同对象）")

print()
print("原始 scores 未受影响 :", scores)
```

### 4-2 写入并读取文本文件

```python
RAW_DATA = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(STUDENTS_FILE, "w", encoding="utf-8") as f:
    f.write(RAW_DATA)

print(f"已写入 {STUDENTS_FILE}")
print("字节数 :", os.path.getsize(STUDENTS_FILE))

print()
print("读取并跳过表头：")
with open(STUDENTS_FILE, "r", encoding="utf-8") as f:
    header = f.readline().strip()
    print("  表头 :", header)
    for lineno, line in enumerate(f, start=2):
        stripped = line.strip()
        if not stripped:
            continue
        parts = stripped.split(",")
        print(f"  第{lineno:>2}行 {stripped!r:<18} → split(',') = {parts}")
```

### 4-3 parse_student(line) 解析与校验函数

校验三件事，任一不满足就抛 `ValueError`：

1. 字段数量必须是 2；
2. 姓名不能为空；
3. 成绩必须能转成整数，且落在 0~100。

注意 `int("abc")` 本身会抛 `ValueError`，这里把它转换成信息更明确的错误再抛出。

```python
def parse_student(line):
    """把 'name,score' 一行解析为 {'name': str, 'score': int}。

    字段数不为 2、姓名为空、成绩非整数或超出 0~100 时抛 ValueError。
    """
    parts = [p.strip() for p in line.strip().split(",")]

    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}：{line.strip()!r}")

    name, raw_score = parts
    if not name:
        raise ValueError(f"姓名为空：{line.strip()!r}")

    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是合法整数：{raw_score!r}") from None

    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")

    return {"name": name, "score": score}


print("正常解析 :", parse_student("Alice,85"))
print("带空格   :", parse_student("  Bob , 58 "))
print("带换行   :", parse_student("Carol,92\n"))
```

### 4-4 逐行处理并捕获非法记录

按行处理文件，用 `try/except ValueError` 捕获非法记录，**记录行号与错误原因后继续处理后续记录**，而不是中断整个程序。

有效记录保存为「字典组成的列表」。

```python
valid_records = []
invalid_records = []

with open(STUDENTS_FILE, "r", encoding="utf-8") as f:
    next(f)                                   # 跳过表头
    for lineno, line in enumerate(f, start=2):
        if not line.strip():
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:
            invalid_records.append({"line": lineno, "raw": line.strip(), "reason": str(exc)})

print(f"有效记录 {len(valid_records)} 条：")
for rec in valid_records:
    print(f"  {rec['name']:<7} {rec['score']:>3}")

print()
print(f"无效记录 {len(invalid_records)} 条：")
for rec in invalid_records:
    print(f"  第{rec['line']:>2}行 {rec['raw']!r:<16} 原因：{rec['reason']}")

print()
print("有效记录类型 :", type(valid_records).__name__, "元素类型:", type(valid_records[0]).__name__)
```

### 4-5 写出统计报告，并演示 w 与 a 模式

先按 `w` 模式写入统计结果，再用 `a` 模式追加一行说明。

- `"w"`（write）：**清空**文件原有内容后从头写；
- `"a"`（append）：保留原有内容，在**末尾追加**。

```python
report_valid = len(valid_records)
report_invalid = len(invalid_records)
report_avg = sum(r["score"] for r in valid_records) / report_valid
report_pass = sum(1 for r in valid_records if r["score"] >= 60) / report_valid

report_lines = [
    "实验01 数据处理报告",
    f"有效记录数：{report_valid}",
    f"无效记录数：{report_invalid}",
    f"平均分：{report_avg:.2f}",
    f"及格率：{report_pass:.2%}",
]
report_text = "\n".join(report_lines) + "\n"

# --- w 模式：覆盖写入 ---
with open(REPORT_FILE, "w", encoding="utf-8") as f:
    f.write(report_text)
print('以 "w" 模式写入后，文件内容：')
print("-" * 40)
print(open(REPORT_FILE, encoding="utf-8").read(), end="")
print("-" * 40)

# --- w 模式再次写入：验证原有内容被清空 ---
with open(REPORT_FILE, "w", encoding="utf-8") as f:
    f.write("第二行写入（w 模式）\n")
print('再用 "w" 模式写一行，之前的内容被清空：')
print(repr(open(REPORT_FILE, encoding="utf-8").read()))

# --- a 模式：追加 ---
with open(REPORT_FILE, "w", encoding="utf-8") as f:
    f.write(report_text)
with open(REPORT_FILE, "a", encoding="utf-8") as f:
    f.write("--- 以上为首次写入，以下为 a 模式追加 ---\n")

print()
print('先 w 后 a，文件内容（原有内容保留并追加）：')
print("-" * 40)
print(open(REPORT_FILE, encoding="utf-8").read(), end="")
print("-" * 40)
print('结论："w" 会清空重写，"a" 在末尾追加且保留原有内容。')
```

### 4-7 保存为 student_utils.py 模块并调用

用 `%%writefile` 把 `parse_student()` 保存到 Notebook 同级的 `student_utils.py` 模块中，然后 `import` 调用，分别展示一个有效输入和一个无效输入的结果。

```python
%%writefile student_utils.py
# -*- coding: utf-8 -*-
"""实验01 用到的学生记录解析工具。由 Notebook 4-7 生成。"""


def parse_student(line):
    """把 'name,score' 一行解析为 {'name': str, 'score': int}。

    字段数不为 2、姓名为空、成绩非整数或超出 0~100 时抛 ValueError。
    """
    parts = [p.strip() for p in line.strip().split(",")]

    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}：{line.strip()!r}")

    name, raw_score = parts
    if not name:
        raise ValueError(f"姓名为空：{line.strip()!r}")

    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是合法整数：{raw_score!r}") from None

    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")

    return {"name": name, "score": score}
```

## 实验5：类、继承与对象复制

### 5-1 定义 Student 类

`__init__()` 初始化 `name` 和 `score`；`is_passed()` 判断是否及格（`score >= 60`）；`__str__()` 决定 `print()` 对象时的显示内容。

```python
class Student:
    """一名学生的姓名与成绩。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩达到 60 分即视为及格。"""
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name!r}, score={self.score}, {status})"


s = Student("Alice", 85)
print("print(obj)      →", s)
print("str(obj)        →", str(s))
print("repr(obj)       →", repr(s))
print("obj.name        →", s.name)
print("obj.score       →", s.score)
print("obj.is_passed() →", s.is_passed())
print()
print("不及格对象 :", Student("Bob", 58))
```

### 5-2 为有效记录创建 Student 对象

遍历实验 4 得到的 `valid_records`，为每条记录创建一个 `Student` 对象，输出对象信息和及格状态。

```python
students = [Student(rec["name"], rec["score"]) for rec in valid_records]

print(f"共创建 {len(students)} 个 Student 对象：")
print()
print(f"{'姓名':<8}{'成绩':>5}   及格状态")
print("-" * 30)
for stu in students:
    passed = "及格" if stu.is_passed() else "不及格"
    print(f"{stu.name:<8}{stu.score:>5}   {passed}")

print()
print("对象列表 :", students)
print()
print("及格人数 :", sum(1 for stu in students if stu.is_passed()), "/", len(students))
```

### 5-3 继承：GraduateStudent

`GraduateStudent` 继承自 `Student`，增加 `research_topic` 属性：

- 用 `super().__init__(name, score)` 初始化父类属性，避免重复代码；
- 重写 `__str__()`，在父类信息基础上补充研究方向。

```python
class GraduateStudent(Student):
    """在 Student 基础上增加研究方向属性的研究生。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)      # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"topic={self.research_topic!r})")


g = GraduateStudent("Carol", 92, "机器学习")
print("print(obj)        →", g)
print("is_passed()       →", g.is_passed(), "（继承自父类，未重写）")
print("research_topic    →", g.research_topic)
print()
print("继承关系检查：")
print("  isinstance(g, GraduateStudent) →", isinstance(g, GraduateStudent))
print("  isinstance(g, Student)         →", isinstance(g, Student))
print("  issubclass(GraduateStudent, Student) →", issubclass(GraduateStudent, Student))
print()
print("父类与子类的 MRO :", [c.__name__ for c in GraduateStudent.__mro__])
```

### 5-4 直接赋值、浅拷贝与深拷贝

对一个**含嵌套列表**的对象 `{"name": "Alice", "scores": [85, 90]}`，比较三种"复制"方式在修改内部列表后的表现：

| 方式 | 外层字典 | 内层列表 |
|---|---|---|
| 直接赋值 `=` | 同一对象 | 同一对象 |
| 浅拷贝 `copy.copy()` | 新对象 | **仍是同一对象** |
| 深拷贝 `copy.deepcopy()` | 新对象 | 新对象 |

> 按要求，每次比较都从**新建的原始对象**开始，避免上一次的修改影响下一次。

```python
import copy


def make_original():
    """每次都新建一个原始对象，保证三组比较互不影响。"""
    return {"name": "Alice", "scores": [85, 90]}


def show(label, original, candidate):
    nested_same = original["scores"] is candidate["scores"]
    outer_same = original is candidate
    print(f"{label}")
    print(f"    原始对象 : {original}")
    print(f"    复制对象 : {candidate}")
    print(f"    外层是同一对象? {outer_same}")
    print(f"    内层列表是同一对象? {nested_same}")
    print()


print("=" * 60)
print("① 直接赋值 alias = original")
print("=" * 60)
original = make_original()
alias = original
alias["scores"].append(100)
show("修改 alias['scores'].append(100) 之后：", original, alias)
print("结论：直接赋值不产生新对象，改 alias 就是改 original。")

print("=" * 60)
print("② 浅拷贝 shallow = copy.copy(original)")
print("=" * 60)
original = make_original()           # 从新建的原始对象开始
shallow = copy.copy(original)
shallow["scores"].append(100)
show("修改 shallow['scores'].append(100) 之后：", original, shallow)
print("结论：浅拷贝只复制外层，嵌套的可变对象仍被共享，所以 original 也被改了。")

print("=" * 60)
print("③ 深拷贝 deep = copy.deepcopy(original)")
print("=" * 60)
original = make_original()           # 从新建的原始对象开始
deep = copy.deepcopy(original)
deep["scores"].append(100)
show("修改 deep['scores'].append(100) 之后：", original, deep)
print("结论：深拷贝递归复制所有层级，改 deep 完全不影响 original。")
```
